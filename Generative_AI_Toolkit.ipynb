### Audio Generation using gtts

```python
from gtts import gTTS    ###simple text to audio..
from IPython.display import Audio


text=input()  #here enter the text  englisg\h

tts=gTTS(text=text,lang='en',tld='co.in',slow=True)
tts.save("o.mp3")  ##mp3
Audio("o.mp3")
```

# Example 2:
## Code Generation using Salesforce/codegen-350M-mono

```python
from transformers import pipeline  ## call we use forcalling the model

generator=pipeline("text-generation",model="Salesforce/codegen-350M-mono")  ## generative model by huging face

prompt="write a program function to check if the number is prime\n"


#Generate code

output=generator(
    prompt,
    max_length=100,
    num_return_sequences=1

)

print(output[0]['generated_text'])
```

# Example 3:

## Traslator model

```python
from transformers import pipeline
translator = pipeline('translation_en_to_hi')
result = translator("Hello, how are you?")
print(result)
```

# Example 4:

## Filling Mask



```python
unmasker = pipeline('fill-mask')
result = unmasker("The man went to the <mask>.")
print(result)
```

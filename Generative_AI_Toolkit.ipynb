# Example 2:
## Code Generation using Salesforce/codegen-350M-mono

```python
from transformers import pipeline  ## call we use forcalling the model

generator=pipeline("text-generation",model="Salesforce/codegen-350M-mono")  ## generative model by huging face

prompt="write a program function to check if the number is prime\n"


#Generate code

output=generator(
    prompt,
    max_length=100,
    num_return_sequences=1

)

print(output[0]['generated_text'])
```

# Example 3:

## Traslator model

```python
from transformers import MarianMTModel, MarianTokenizer

# Load the tokenizer and model directly
model_name = "Helsinki-NLP/opus-mt-en-hi"
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

# Translate the text
text = "Hello, how are you?"
tokenized_text = tokenizer(text, return_tensors="pt", padding=True)
translation = model.generate(**tokenized_text)
result = tokenizer.decode(translation[0], skip_special_tokens=True)

print(result)
```
